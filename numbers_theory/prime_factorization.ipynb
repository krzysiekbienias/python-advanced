Write a function prime_factors(n) that returns a list of all prime factors of a positive integer 
n
n, including duplicates (e.g., prime_factors(12) returns [2, 2, 3]).


```python
import math
```

```python

def prime_factors(n):
    res=[]
    if n<=1:
        return []
    while (n%2==0):
        res.append(2)
        n/=2
    p=3
    while p*p<=n:
        while n%p==0:
            res.append(p)
            n/=p
        p+=2
    if n>1:
        res.append(n)
        
    return res
```
