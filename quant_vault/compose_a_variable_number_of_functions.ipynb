Write a function `compose_apply(ops, x)` that applies a sequence of single-argument operations to an integer x in right-to-left order and returns the resulting value.

`ops` is a list of `[op_name, k]` specs. The LAST spec in the list is applied first to `x`, its result is fed into the previous spec, and so on, until the first spec produces the final output. This is ordinary mathematical composition: for ops [f, g, h] the result equals f(g(h(x))).

Operations

* ["add", k] -> x + k
* ["sub", k] -> x - k
* ["mul", k] -> x * k
* ["neg", 0] -> -x (the k is ignored)
* ["square", 0] -> x * x (the k is ignored)
Constraints
Any number of ops may be given, including zero.
If ops is empty, return x unchanged (identity).
Return

Return the single final value (an integer), not the list of ops or intermediate results.

```python
def perf_operation(currenc_val,op,new_val):
    if op=="add":
        return currenc_val+new_val
    if op=="sub":
        return currenc_val-new_val
    if op=="mul":
        return currenc_val*new_val
    if op=="neg":
        return -1*currenc_val 
    if op=="square":
        return currenc_val*currenc_val   
```

### Solution from answer

```python
def compose_apply(ops, x):
    # ops is a list of [op_name, k] specs. Build the composed function
    # (right-to-left), apply it to x, and return the result.
    def make(op_name, k):
        if op_name == "add":
            return lambda v: v + k
        if op_name == "sub":
            return lambda v: v - k
        if op_name == "mul":
            return lambda v: v * k
        if op_name == "neg":
            return lambda v: -v
        if op_name == "square":
            return lambda v: v * v
        raise ValueError("unknown op: " + str(op_name))

    funcs = [make(name, k) for name, k in ops]
    for f in reversed(funcs):
        x = f(x)
    return x
```
