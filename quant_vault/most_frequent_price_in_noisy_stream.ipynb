You are given a list of a stock's price lines (one raw string value per element). Some lines are corrupt because the feed comes from an unreliable external source. Implement:

most_frequent_price(lines)
where lines is a list of strings. Find the price that occurs most often and how many times it occurred.

A price is valid only if it parses as a positive, non-NaN number (for example 6.745, 1.32, 1.2, and 5 are all valid). Anything that does not parse, is NaN, or is zero or negative must be ignored. Blank/whitespace-only lines are ignored.

Return (do not print) a string: the winning frequency, then a colon :, then a comma-separated list (ascending) of all values that hit that frequency. If several values tie for the top count, list them all. Whole numbers are formatted without a trailing .0 (so 3.0 and 3 count as the same value 3). If there are no valid prices at all, return the default "0:".

````python
from collections import Counter
import math
````

````python
def is_valid_price(price):
    price=price.strip()
    if not price:
        return False
    try:
        price_f=float(price)
    except ValueError:
        return False
    return math.isfinite(price_f) and price_f>0
````

````python
def fmt(x):
    return str(int(x)) if x == int(x) else str(x)
````

````python

def most_frequent_price(lines):
    cleaned_stream=[float(p)  for p in lines if is_valid_price(p)]
    di=Counter(cleaned_stream)    
    max_di=max(di.values())
    
    corresponding_keys=[k for k,v in di.items() if v==max_di]
    corresponding_keys.sort(key=float)
    res=str(max_di)+":"
    for i in range(len(corresponding_keys)-1):
        res+=corresponding_keys[i]+','

    return res+corresponding_keys[-1]

````

````python
def most_frequent_price(lines):
    cleaned = [float(p) for p in lines if is_valid_price(p)]
    if not cleaned:
        return "0:"
    di = Counter(cleaned)
    top = max(di.values())
    keys = sorted(k for k, v in di.items() if v == top)
    return f"{top}:{','.join(fmt(k) for k in keys)}"
````
