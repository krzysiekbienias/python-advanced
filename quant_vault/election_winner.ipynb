You are given a list of names, one per vote (so the same name can appear many times). Return the winner of the election, defined as the name with the highest vote count. If several names tie for the highest count, the winner is the one that comes last in alphabetical order.

```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
from collections import Counter


def election_winner(votes):
    di=Counter(votes)
    
    return max (di,key =lambda name:(di[name],name))
    
```

```python
# testowe inputy — odpalaj election_winner(...) po napisaniu funkcji

votes_clear = ["Alice", "Bob", "Alice", "Alice"]
# Alice: 3, Bob: 1  → Alice

votes_tie_alpha = ["Bob", "Alice", "Bob", "Alice"]
# Alice: 2, Bob: 2  → Bob (później alfabetycznie)

votes_three_way = ["Ann", "Ben", "Cara", "Ann", "Ben", "Cara"]
# wszyscy po 2  → Cara

votes_single = ["Zoe"]
# → Zoe

votes_all_same = ["Mia", "Mia", "Mia"]
# → Mia

votes_later_name_fewer = ["Zack", "Amy", "Amy"]
# Amy: 2, Zack: 1  → Amy (więcej głosów wygrywa, alfabet nie wchodzi)

votes_clear
election_winner(votes_clear)
election_winner(votes_tie_alpha)
election_winner(votes_three_way)
election_winner(votes_single)
election_winner(votes_all_same)
election_winner(votes_later_name_fewer)
```
